implemenatasikan inverse citra padaulasan teori dengan rumus


```python
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow

print('Inverse Citra')
print('-------------')

# Sesuaikan dengan lokasi file gambar kamu
original = cv.imread('/content/drive/MyDrive/PCDVK2026/peppers.jpg')

# Membuat array kosong dengan ukuran dan tipe data yang sama
inverse_image = np.zeros(original.shape, original.dtype)

# Akses setiap piksel
for y in range(original.shape[0]):
    for x in range(original.shape[1]):
        for c in range(original.shape[2]):
            # Rumus inverse: 255 - nilai piksel asli
            inverse_image[y, x, c] = 255 - original[y, x, c]

# Menggabungkan citra asli dan citra inverse
final_frame = cv.hconcat((original, inverse_image))

# Menampilkan hasil
cv2_imshow(final_frame)
```

mplementasi transformasi contrast

```python
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow

print('Mengubah kontras dan tingkat kecerahan citra')
print('--------------------------------------------')

brightness = int(input('Masukkan tingkat kecerahan: '))
contrast = float(input('Masukkan kontras: '))

original = cv.imread('/content/drive/MyDrive/PCDVK2026/sunset.jpg')


brightness_image = np.zeros(original.shape, dtype=np.uint8)

for y in range(original.shape[0]):
    for x in range(original.shape[1]):
        for c in range(original.shape[2]):

            brightness_image[y, x, c] = np.clip(
                int(original[y, x, c]) + brightness,
                0,
                255
            )


F = contrast

contrast_image = np.zeros(original.shape, dtype=np.uint8)

for y in range(original.shape[0]):
    for x in range(original.shape[1]):
        for c in range(original.shape[2]):

            pixel = float(original[y, x, c])

            # Rumus g(x,y) = a * f(x,y) + b
            new_pixel = contrast * pixel + brightness

            # Truncate nilai 0-255
            contrast_image[y, x, c] = np.clip(
                new_pixel,
                0,
                255
            )


final_frame = cv.hconcat((original, contrast_image))

cv2_imshow(final_frame)
```

Implementasikan transformasi logarithmic brightness pada Google Colaboratory
menggunakan formula yang terdapat pada bagian Ulasan Teori untuk transformasi
log, sehingga menghasilkan keluaran seperti berikut:

```python
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow

print('Mengubah tingkat kecerahan citra dengan Transformasi Log')
print('--------------------------------------------------------')

c = float(input('Masukkan nilai kecerahan: '))

original = cv.imread('/content/drive/MyDrive/PCDVK2026/houses.jpg')

log_image = np.zeros(original.shape, dtype=np.uint8)

for y in range(original.shape[0]):
    for x in range(original.shape[1]):
        for ch in range(original.shape[2]):

            r = float(original[y, x, ch])

            # Rumus Transformasi Log
            s = c * np.log(1 + r)

            log_image[y, x, ch] = np.clip(s, 0, 255)

final_frame = cv.hconcat((original, log_image))

cv2_imshow(final_frame)
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

original = cv.imread('/content/drive/MyDrive/PCDVK2026/peppers.jpg')

original_rgb = cv.cvtColor(original, cv.COLOR_BGR2RGB)

grayscale_avg = np.zeros(original.shape[:2], dtype=np.uint8)

for y in range(original.shape[0]):
    for x in range(original.shape[1]):

        R = float(original_rgb[y, x, 0])
        G = float(original_rgb[y, x, 1])
        B = float(original_rgb[y, x, 2])

        grayscale_avg[y, x] = np.clip((R + G + B) / 3, 0, 255)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(original_rgb)

plt.subplot(1, 2, 2)
plt.imshow(grayscale_avg, cmap='gray')

plt.suptitle('a.  Averaging', x=0.2, y=0.98, fontsize=14)

plt.tight_layout()
plt.show()
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

original = cv.imread('/content/drive/MyDrive/PCDVK2026/peppers.jpg')
original_rgb = cv.cvtColor(original, cv.COLOR_BGR2RGB)

hsv = cv.cvtColor(original, cv.COLOR_BGR2HSV)

lower_red1 = np.array([0, 50, 50])
upper_red1 = np.array([10, 255, 255])
lower_red2 = np.array([170, 50, 50])
upper_red2 = np.array([180, 255, 255])

mask1 = cv.inRange(hsv, lower_red1, upper_red1)
mask2 = cv.inRange(hsv, lower_red2, upper_red2)
red_mask = mask1 | mask2

gray = cv.cvtColor(original_rgb, cv.COLOR_RGB2GRAY)
gray_3channel = cv.cvtColor(gray, cv.COLOR_GRAY2RGB)


result = np.where(red_mask[:, :, np.newaxis] == 255, original_rgb, gray_3channel)


plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(original_rgb)


plt.subplot(1, 2, 2)
plt.imshow(result)

plt.tight_layout()
plt.show()
```

gamma correction

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

print(' Gamma Correction pada citra ')
print('-----------------------------------')
try:
    gamma = float(input('Masukkan nilai Gamma: '))
except ValueError:
    print('Error, not a number')
    gamma = 1.0

original = cv.imread('/content/drive/MyDrive/PCDVK2026/houses.jpg')
original_rgb = cv.cvtColor(original, cv.COLOR_BGR2RGB)

gamma_image = np.array(255 * ((original_rgb / 255.0) ** gamma), dtype=np.uint8)


plt.figure(figsize=(12, 6))


plt.subplot(1, 2, 1)
plt.imshow(original_rgb)
plt.title('Citra Asli')
plt.axis('off')


plt.subplot(1, 2, 2)
plt.imshow(gamma_image)
plt.title(f'Gamma Correction ($\gamma$ = {gamma})')
plt.axis('off')

plt.tight_layout()
plt.show()


print(f"Gambar diatas adalah citra yang diproses menggunakan Gamma Correction nilai gamma {gamma}.")
```

simulasi depth

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt


print(' Kuantisasi Citra Grayscale (Bit Depth Reduction) ')
print('--------------------------------------------------')
try:
    bit_depth = int(input('Masukkan kedalaman bit_depth (1-8): '))
except ValueError:
    print('Error, not a number')
    bit_depth = 8


original = cv.imread('/content/drive/MyDrive/PCDVK2026/peppers.jpg')
gray_image = cv.cvtColor(original, cv.COLOR_BGR2GRAY)

level = 255 / (2**bit_depth - 1)

quantized_image = np.round(gray_image / level) * level
quantized_image = np.clip(quantized_image, 0, 255).astype(np.uint8)

plt.figure(figsize=(12, 6))


plt.subplot(1, 2, 1)
plt.imshow(gray_image, cmap='gray')
plt.title('Citra Grayscale (8-bit)')
plt.axis('off')


plt.subplot(1, 2, 2)
plt.imshow(quantized_image, cmap='gray')
plt.title(f'Kuantisasi Grayscale ({bit_depth}-bit)')
plt.axis('off')

plt.tight_layout()
plt.show()

print(f"Gambar diatas adalah hasil kuantisasi grayscale dengan kedalaman {bit_depth}-bit ({2**bit_depth} tingkat keabu-aabuan).")
```

Buat modul Average Denoising

```python
import cv2 as cv
import numpy as np
import glob
import math
import matplotlib.pyplot as plt


def calculate_psnr(original, denoised):
    mse = np.mean((original.astype(np.float64) - denoised.astype(np.float64)) ** 2)
    if mse == 0:
        return float('inf')
    max_pixel = 255.0
    psnr = 20 * math.log10(max_pixel / math.sqrt(mse))
    return psnr


original = cv.imread('/content/drive/MyDrive/PCDVK2026/galaxy.jpg')

cv_img = []
for img in sorted(glob.glob('/content/drive/MyDrive/PCDVK2026/noises/*.jpg')):
    n = cv.imread(img)
    cv_img.append(n)


list_N = [5, 10, 20, 50, 100]
results_psnr = {}

for N in list_N:
    subset_imgs = cv_img[:N]

    avg_img = np.mean(subset_imgs, axis=0)
    avg_img = np.clip(avg_img, 0, 255).astype(np.uint8)

    psnr_value = calculate_psnr(original, avg_img)
    results_psnr[N] = psnr_value
    print(f"Jumlah Citra (N = {N:3d}) | Nilai PSNR: {psnr_value:.4f} dB")

plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.imshow(cv.cvtColor(original, cv.COLOR_BGR2RGB))
plt.title('Citra Asli (galaxy.jpg)')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cv.cvtColor(avg_img, cv.COLOR_BGR2RGB))
plt.title(f'Average Denoising (N = 100)\nPSNR: {results_psnr[100]:.2f} dB')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

# 1. Baca Citra Asli
img_bgr = cv.imread('/content/drive/MyDrive/PCDVK2026/couple.tiff')
img = cv.cvtColor(img_bgr, cv.COLOR_BGR2RGB)
h, w, c = img.shape

# 2. Buat Mask (2 Lingkaran Putih pada Latar Hitam)
mask = np.zeros((h, w, c), dtype=np.uint8)

# Posisi titik pusat & radius lingkaran disesuaikan dengan wajah/tubuh subjek
radius = int(min(h, w) * 0.18)
center1 = (int(w * 0.25), int(h * 0.22))  # Lingkaran kiri
center2 = (int(w * 0.75), int(h * 0.22))  # Lingkaran kanan

cv.circle(mask, center1, radius, (255, 255, 255), -1)
cv.circle(mask, center2, radius, (255, 255, 255), -1)
```

not

```python
# Operator NOT (membalikkan nilai piksel mask)
mask_not = cv.bitwise_not(mask)

# Tampilkan Hasil
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1); plt.imshow(mask); plt.title('Input (Mask)'); plt.axis('off')
plt.subplot(1, 2, 2); plt.imshow(mask_not); plt.title('Output (NOT)'); plt.axis('off')
plt.show()
```

```python
# Operator OR antara Citra Asli dan Mask
out_or = cv.bitwise_or(img, mask)

# Tampilkan Hasil
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1); plt.imshow(img); plt.title('Input (Citra Asli)'); plt.axis('off')
plt.subplot(1, 2, 2); plt.imshow(out_or); plt.title('Output (OR)'); plt.axis('off')
plt.show()
```

```python
# Operator AND antara Citra Asli dan Mask
out_and = cv.bitwise_and(img, mask)

# Tampilkan Hasil
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1); plt.imshow(img); plt.title('Input (Citra Asli)'); plt.axis('off')
plt.subplot(1, 2, 2); plt.imshow(out_and); plt.title('Output (AND)'); plt.axis('off')
plt.show()
```

```python
# Operator NAND (Inversi dari hasil AND)
out_nand = cv.bitwise_not(cv.bitwise_and(img, mask))

# Tampilkan Hasil
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1); plt.imshow(img); plt.title('Input (Citra Asli)'); plt.axis('off')
plt.subplot(1, 2, 2); plt.imshow(out_nand); plt.title('Output (NAND)'); plt.axis('off')
plt.show()
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

path_img = '/content/drive/MyDrive/PCDVK2026/couple.tiff'
img_bgr = cv.imread(path_img)

if img_bgr is None:
    raise FileNotFoundError(f"File tidak ditemukan di path: {path_img}. Pastikan nama file dan foldernya benar.")

img = cv.cvtColor(img_bgr, cv.COLOR_BGR2RGB)
h, w, c = img.shape

mask_circle_rgb = np.zeros((h, w, c), dtype=np.uint8)

radius = int(min(h, w) * 0.18)
center1 = (int(w * 0.25), int(h * 0.22))
center2 = (int(w * 0.75), int(h * 0.22))

cv.circle(mask_circle_rgb, center1, radius, (255, 255, 255), -1)
cv.circle(mask_circle_rgb, center2, radius, (255, 255, 255), -1)

masked_img = cv.bitwise_and(img, mask_circle_rgb)
inv_mask_circle_rgb = cv.bitwise_not(mask_circle_rgb)
white_bg = np.full((h, w, c), 255, dtype=np.uint8)
only_bg_white = cv.bitwise_and(white_bg, inv_mask_circle_rgb)


final_result = cv.add(masked_img, only_bg_white)


plt.figure(figsize=(15, 6))

plt.subplot(1, 3, 1)
plt.imshow(img)
plt.title('Citra Asli (couple.tiff)')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(mask_circle_rgb, cmap='gray')
plt.title('Masking (2 Lingkaran Putih)')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(final_result)
plt.title('Hasil Akhir (Latar Putih Pekat)')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt


path_img = '/content/drive/MyDrive/PCDVK2026/crayfish.jpg'
original = cv.imread(path_img)

if original is None:
    raise FileNotFoundError(f"File tidak ditemukan di path: {path_img}.")

original_rgb = cv.cvtColor(original, cv.COLOR_BGR2RGB)

C = 90.0

F = (259.0 * (C + 255.0)) / (255.0 * (259.0 - C))


contrast_rgb = F * (original_rgb.astype(np.float32) - 128.0) + 128.0


contrast_rgb = np.clip(contrast_rgb, 0, 255).astype(np.uint8)

plt.figure(figsize=(14, 7))

plt.subplot(1, 2, 1)
plt.imshow(original_rgb)
plt.title('Before: Citra Asli')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(contrast_rgb)
plt.title(f'After: Contrast Transformation (C = {C}, F = {F:.2f})')
plt.axis('off')

plt.tight_layout()
plt.show()
```
